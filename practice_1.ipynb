```python
# Check Whether a Number is a Perfect Square

import math

n = int(input("Enter the Number : "))

root = math.isqrt(n)

if root * root == n:
    print("Perfect Square")
else:
    print("Not a Perfect Square")
```

```python
# Format Date and Time

from datetime import datetime
from zoneinfo import ZoneInfo

india_time = datetime.now(ZoneInfo("Asia/Kolkata"))

print(india_time.strftime("%d-%m-%Y"))
print(india_time.strftime("%d-%m-%y"))

print(india_time.strftime("%d-%m-%Y %I:%M:%S %p"))
```

```python
# Indian Date and Time Formatting

from datetime import datetime
from zoneinfo import ZoneInfo

india_time = datetime.now(ZoneInfo("Asia/Kolkata"))

print("Indian Date : ", india_time.strftime("%d/%m/%y"))
print("Indian Time 12hrs : ", india_time.strftime("%I:%M:%S %p"))
print("Indian Time 24hrs : ", india_time.strftime("%H:%M:%S"))
```

```python
# datetime() - Create Custom Date and Time

from datetime import datetime

dt = datetime(1997, 12, 9, 4, 30, 45)

print(dt)
```

```python
# Date Attributes and Weekday

from datetime import datetime
from zoneinfo import ZoneInfo

india_time = datetime.now(ZoneInfo("Asia/Kolkata"))

print(india_time.year)
print(india_time.month)
print(india_time.day)

# weekday() - Monday = 0, Sunday = 6
print(india_time.weekday())

# isoweekday() - Monday = 1, Sunday = 7
print(india_time.isoweekday())
```

```python
# replace() - Change the Year

from datetime import datetime
from zoneinfo import ZoneInfo

india_time = datetime.now(ZoneInfo("Asia/Kolkata"))

print(india_time)

new_dt = india_time.replace(year=2030)

print(new_dt)
```

```python
# timedelta() - Add One Day

from datetime import datetime, timedelta

today = datetime.now()

tomorrow = today + timedelta(days=1)

print(today)
print(tomorrow)
```

```python
from datetime import date, timedelta

today = date.today()
print(today)

future_date = today + timedelta(days=10)

print("After 10 Days:", future_date)
```

```python
from datetime import date, timedelta

today = date.today()
print(today)

past_date = today - timedelta(days=10)

print("Before 10 Days:", past_date)
```

```python
from datetime import date

D1 = date(1997, 12, 9)
D2 = date(2026, 9, 30)

difference = D2 - D1

print(difference.days)
```

```python
import calendar

year = int(input("Enter the Year: "))

if calendar.isleap(year):
    print(year, ": Leap Year")
else:
    print(year, ": Not Leap Year")
```
