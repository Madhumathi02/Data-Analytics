```python
#Filtering data-Find employees whose salary is greater than 60,000
answer=df[df ["Salary"]>60000]
print(answer)
```

```python
#Multiple conditions
#AND
answer=df[
    (df["Salary"]>7000) &
    (df["Department"]=="IT")
]
print(answer)
```

```python
#Multiple conditions
#AND
answer=df[
    (df["Salary"]>7000) &
    (df["Department"]=="Marketing")
]
print(answer)
```

```python
#OR

answer=df[
    (df["Department"]=="IT") |
    (df["Department"]=="HR")
]
print(answer)
```

```python
#isin()-check whether values belong to a list

answer=df[df["Department"].isin(["Marketing","HR","Sales"])]
print(answer)
```

```python
#between()-Filter the values within a range

answer=df[df["Salary"].between(75000,80000)]
print(answer)
```

```python
#Total Salary-department ooda total salary
answer=df.groupby("Department")["Salary"].sum()
print(answer)
```

```python
answer=df.groupby("Department")["Salary"].min()  #minimum salary in each depart
print(answer)
```

```python
answer=df.groupby("City")["Salary"].min()
print(answer)
```

```python
#max()
answer=df.groupby("City")["Salary"].max()
print(answer)
```

```python
#count()
answer=df.groupby("City")["Salary"].count()
print(answer)
```

```python
#agg()-performs multiple calculations
answer=df.groupby("Department")["Salary"].agg(["sum","mean","min","max","count"])
print(answer)
```

```python
#Creating a new column
df["AnnualSalary"]= df["Salary"]*12
print(df)
```

```python
#apply()-applies fn to data
def Salary_grade(Salary):
  if(Salary>=6000):
    return "A"
  elif (Salary>=5000):
    return "B"
  else:
    return "C"
df["Grade"]=df["Salary"].apply(Salary_grade)
print(df)
```

```python
#query()-Filters DataFrame using a query expression
answer=df.query("Salary>=6000")
print(answer)
```

```python
#Multiple conditions
answer=df.query("Salary>=70000 and Experience>=10")
print(answer)
```
