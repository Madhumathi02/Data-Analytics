```python
#coding test
# 1. List - Largest, Smallest, Sum and Average

numbers = []

for i in range(10):
    n = int(input("Enter a number: "))
    numbers.append(n)

print("Largest:", max(numbers))
print("Smallest:", min(numbers))
print("Sum:", sum(numbers))
print("Average:", sum(numbers) / len(numbers))
```

```python
# 2. List - Student Name Operations

names = []

for i in range(10):
    name = input("Enter student name: ")
    names.append(name)

print("Student Names:", names)

names.append(input("Enter name to add: "))
print("After adding:", names)

old = input("Enter name to update: ")
if old in names:
    new = input("Enter new name: ")
    index = names.index(old)
    names[index] = new
print("After updating:", names)

remove_name = input("Enter name to remove: ")
if remove_name in names:
    names.remove(remove_name)
print("After removing:", names)

search_name = input("Enter name to search: ")
if search_name in names:
    print("Name found")
else:
    print("Name not found")

print("Final List:", names)
```

```python
# 3. List - Even Numbers

numbers = []

n = int(input("How many numbers? "))

for i in range(n):
    numbers.append(int(input("Enter a number: ")))

even = []

for num in numbers:
    if num % 2 == 0:
        even.append(num)

print("Original List:", numbers)
print("Even Numbers:", even)
```

```python
# 4. List - Remove Duplicates

numbers = [10, 20, 10, 30, 20, 40, 30]
unique = []

for num in numbers:
    if num not in unique:
        unique.append(num)

print("Original List:", numbers)
print("List without duplicates:", unique)
```

```python
# 5. List - Second Largest Number

numbers = [10, 50, 30, 80, 60, 80]

largest = None
second = None

for num in numbers:
    if largest is None or num > largest:
        if num != largest:
            second = largest
        largest = num
    elif num != largest and (second is None or num > second):
        second = num

if second is not None:
    print("Second Largest:", second)
else:
    print("No second-largest distinct number")
```

```python
# 6. List - Frequency of Each Element

numbers = [10, 20, 10, 30, 20, 10]
frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

print("Frequency:", frequency)
```

```python
# 10. Tuple - Convert, Modify and Convert Back

numbers = (10, 20, 30, 40)

lst = list(numbers)
print("Original List:", lst)

lst[1] = 25
lst.append(50)

numbers = tuple(lst)

print("Modified List:", lst)
print("Final Tuple:", numbers)
```

```python
# 11. Set - Add, Remove, Search and Display

numbers = {10, 20, 30, 40}

print("Original Set:", numbers)

numbers.add(50)
print("After Adding:", numbers)

numbers.discard(20)
print("After Removing:", numbers)

n = int(input("Enter a number to search: "))

if n in numbers:
    print("Element found")
else:
    print("Element not found")

print("Final Set:", numbers)
```

```python
# 12. Set - Set Operations

a = {"Ravi", "Priya", "Arun"}
b = {"Priya", "Arun", "Kumar"}

print("Union:", a | b)
print("Intersection:", a & b)
print("Difference:", a - b)
print("Symmetric Difference:", a ^ b)
```

```python
# 13. Set - Remove Duplicate Values

numbers = [10, 20, 10, 30, 20, 40]

unique = set(numbers)

print("Original List:", numbers)
print("Unique Values:", unique)
```

```python
# 14. Set - Check Subset

skills1 = {"Python", "SQL"}
skills2 = {"Python", "SQL", "Excel", "Power BI"}

if skills1.issubset(skills2):
    print("It is a subset")
else:
    print("It is not a subset")
```

```python
# 16. Dictionary - Student Management

students = {
    101: {"name": "Ravi", "course": "Python", "fees": 5000},
    102: {"name": "Priya", "course": "SQL", "fees": 4000}
}

print("Original Dictionary:", students)

students[103] = {
    "name": "Arun",
    "course": "Java",
    "fees": 4500
}
print("After Adding:", students)

students[101]["fees"] = 5500
print("After Updating:", students)

del students[102]
print("After Deleting:", students)

sid = int(input("Enter student ID to search: "))

if sid in students:
    print("Student Details:", students[sid])
else:
    print("Student not found")

print("Final Dictionary:", students)
```

```python
# 17. Dictionary - Employee Salary

employees = {
    "Ravi": 30000,
    "Priya": 45000,
    "Arun": 25000,
    "Kumar": 40000
}

highest = max(employees, key=employees.get)
lowest = min(employees, key=employees.get)

print("Highest Salary Employee:", highest)
print("Highest Salary:", employees[highest])

print("Lowest Salary Employee:", lowest)
print("Lowest Salary:", employees[lowest])
```

```python
# 18. Dictionary - Word Frequency

sentence = input("Enter a sentence: ")

words = sentence.lower().split()
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print("Word Frequency:", frequency)
```

```python
# 19. Dictionary - Character Frequency

s = input("Enter a string: ")
frequency = {}

for ch in s:
    if ch in frequency:
        frequency[ch] += 1
    else:
        frequency[ch] = 1

print("Character Frequency:", frequency)
```

```python
# 20. Employee Management System

employees = []
employee_ids = set()

# Add Employees
n = int(input("Enter number of employees: "))

for i in range(n):
    emp_id = int(input("Enter employee ID: "))
    name = input("Enter employee name: ")
    department = input("Enter department: ")
    salary = float(input("Enter salary: "))

    employee = {
        "ID": emp_id,
        "Name": name,
        "Department": department,
        "Salary": salary
    }

    employees.append(employee)
    employee_ids.add(emp_id)

# Display All Employees
print("\nAll Employee Records:")

for emp in employees:
    print(emp)

# Search Employee
search_id = int(input("\nEnter employee ID to search: "))

found = False

for emp in employees:
    if emp["ID"] == search_id:
        print("Employee Found:", emp)
        found = True
        break

if not found:
    print("Employee not found")

# Update Employee
update_id = int(input("\nEnter employee ID to update: "))

for emp in employees:
    if emp["ID"] == update_id:
        emp["Name"] = input("Enter new name: ")
        emp["Department"] = input("Enter new department: ")
        emp["Salary"] = float(input("Enter new salary: "))
        print("Employee Updated")
        break
else:
    print("Employee not found")

# Remove Employee
remove_id = int(input("\nEnter employee ID to remove: "))

for emp in employees:
    if emp["ID"] == remove_id:
        employees.remove(emp)
        employee_ids.discard(remove_id)
        print("Employee Removed")
        break
else:
    print("Employee not found")

# Calculate Salary Details
if employees:
    salaries = [emp["Salary"] for emp in employees]

    print("\nHighest Salary:", max(salaries))
    print("Lowest Salary:", min(salaries))
    print("Total Salary:", sum(salaries))

# Identify Unique Departments
departments = set()

for emp in employees:
    departments.add(emp["Department"])

print("Unique Departments:", departments)

# Convert Employee Records into Tuples
employee_tuples = []

for emp in employees:
    record = (
        emp["ID"],
        emp["Name"],
        emp["Department"],
        emp["Salary"]
    )
    employee_tuples.append(record)

print("\nEmployee Records as Tuples:")

for record in employee_tuples:
    print(record)

# Display Final Employee Records
print("\nFinal Employee Records:")

for emp in employees:
    print(emp)
```
