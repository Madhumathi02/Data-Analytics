```python
#Match statement
num1 = int(input("Enter the Num 1 : "))
num2 = int(input("Enter the Num 2 : "))

process = input("Enter the Value +, -, , /, * : ")

match(process):
    case "+":
        print(num1 + num2)
    case "-":
        print(num1 - num2)
    case "*":
        print(num1 * num2)
    case "/":
        print(num1 / num2)
    case "**":
        print(num1 ** num2)
    case _:
        print("You have entered a wrong value. Please enter +, -, , /, *")
```

```python
# Find Largest Number

numbers = [10, 50, 20, 80, 30]

largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num
        print("Step : ", num, " : ", largest)

print("Largest : ", largest)
```

```python
# Find Smallest Number

numbers = [10, 20, 30, 40, 50]

smallest = numbers[0]

for num in numbers:
    if num < smallest:
        smallest = num
        print("Step : ", num, " : ", smallest)

print("Smallest : ", smallest)
```

```python
# Search for a Number

numbers = [10, 20, 30, 40, 50]
search = 30

for num in numbers:
    if num == search:
        print("Number Found : ", num)
        break
else:
    print("Number Not Found")
```

```python
# Prime Number Check

num = int(input("Enter the Number : "))

if num > 1:
    for i in range(2, num):
        if num % i == 0:
            print("Not a Prime Number")
            break
    else:
        print("Prime Number")
else:
    print("Not a Prime Number")
```

```python
# Find Factorial of a Number

num = int(input("Enter the Number : "))

factorial = 1

for i in range(1, num + 1):
    factorial *= i
    # factorial = factorial * i
    print("Step : ", i, " : ", factorial)

print("Factorial : ", factorial)
```

```python
# Reverse a String

text = input("Enter a string : ")

reverse = ""

for ch in text:
    reverse = ch + reverse

print("Reverse : ", reverse)
```
