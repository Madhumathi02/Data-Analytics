```python
 #Retail Transaction Engine

 def calculate_total(price,quantity,tax,shipping):
   sub_total=price*quantity
   tax_amount=sub_total*tax/100

   if sub_total>=1000:
    shipping_fee=0
   else:
    shipping_fee=shipping

   total=sub_total+tax_amount+shipping_fee
   return total

price=int(input("Enter item price:"))
quantity=int(input("Enter quantity:"))
tax=int(input("Enter tax percentage:"))
shipping=int(input("Enter shipping fee:"))

total=calculate_total(price,quantity,tax,shipping)
print("Total",total)
```

```python
# Program 1: Print vs Return - Area of Rectangle

def area_print(length, breadth):
    print("Area:", length * breadth)

def area_return(length, breadth):
    return length * breadth

result1 = area_print(5, 4)
result2 = area_return(5, 4)

print("Result1:", result1)
print("Result2:", result2)
print("Result2 multiplied by 2:", result2 * 2)
```

```python
# Program 2: Profile Builder - Positional and Keyword Arguments

def profile(name, age, country):
    print("Name:", name)
    print("Age:", age)
    print("Country:", country)
    print()

# Positional arguments
profile("Madhu", 23, "India")

# Keyword arguments
profile(name="Arun", age=25, country="India")

# Mixed arguments
profile("Divya", age=22, country="India")
```

```python
# Program 4: Calculate Discount with Error Handling

def calculate_discount(price, discount_pct):
    try:
        if not isinstance(price, (int, float)) or not isinstance(discount_pct, (int, float)):
            print("Error: Inputs must be numbers")
            return

        if discount_pct < 0:
            print("Error: Discount cannot be negative")
            return

        final_price = price - (price * discount_pct / 100)
        print("Final Price:", final_price)

    except TypeError:
        print("Error: Invalid input")

calculate_discount(1000, 10)
calculate_discount(500, -5)
calculate_discount("abc", 10)
```
