```python
#CELSIUS TO FAHRENHEIT
#F=(C*9/5)+32
def celsius_to_fahrenheit(celsius):
  fahrenheit=(celsius*9/5)+32
  return fahrenheit
celsius=int(input("Enter Celsius value:"))
result=celsius_to_fahrenheit(celsius)
print("Fahrenheit:",result)
```

```python
 #Retail Transaction Engine

 def calculate_total(price,quantity,tax,shipping):
   sub_total=price*quantity
   tax_amount=sub_total*tax/100

   if sub_total>=1000:
    shipping_fee=0
   else:
    shipping_fee=shipping

   total=sub_total+tax_amount+shipping_fee
   return total

price=int(input("Enter item price:"))
quantity=int(input("Enter quantity:"))
tax=int(input("Enter tax percentage:"))
shipping=int(input("Enter shipping fee:"))

total=calculate_total(price,quantity,tax,shipping)
print("Total",total)
```
